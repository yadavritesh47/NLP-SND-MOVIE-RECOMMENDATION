```python
import pandas as pd
import joblib
```

```python
df = pd.read_csv(r"C:\Users\suman\OneDrive\Desktop\ML\unsupervised\movies_content.csv")
df
```

```python
df2=df[['movie_id','description','language','director','cast','genre','name']]
df2
```

```python
df3 = df2.dropna()
df4 = df3.reset_index(drop=True)
df4
```

```python
df5 = df4[["movie_id","name"]]
df5
```

```python
df5.loc[:, 'tag'] = df4["description"] + " " + df4["language"] + " " + df4["director"] + " " + df4["cast"] + " " + df4["genre"]
df5
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
tv=TfidfVectorizer(lowercase=True)
vectors=tv.fit_transform(df5.tag).toarray()
vectors.shape
```

```python
from sklearn.neighbors import NearestNeighbors
```

```python
model=NearestNeighbors(metric='cosine')
model.fit(vectors,vectors)
```

```python
name="Andaz Apna Apna"
index=df5[df5.name==name].index[0]
distances,indexes=model.kneighbors([vectors[index]],n_neighbors=10)


# # Normalize both user input and DataFrame column
# name = input("enter movie")  # user input
# normalized_name = name.strip().lower()

# # Create a normalized column if not already
# df5.loc[:, 'normalized_name'] = df5['name'].str.strip().str.lower()

# # Find index using normalized matching
# index = df5[df5['normalized_name'] == normalized_name].index[0]


# # Now use index as before
# distances, indexes = model.kneighbors([vectors[index]], n_neighbors=10)
```
